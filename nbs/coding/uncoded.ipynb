# No channel code

> Bits sent as they are, and detected one by one: the uncoded pair of stages.

Channel coding protects the bits against the channel's errors: an encoder adds redundancy (`bits` → `codeword`),
a decoder uses it to correct errors (`llr` → `bits_hat`). The two are stages of a link chain (see
`radio.link.chain`), wired by these names.

Without a code, the codeword is the bits, and each bit is decided from its own LLR: `NoCoding` and
`HardDecision`. LLRs follow Sionna's sign: `log p(1) / p(0)`, so a positive LLR means 1.

```python
#| export
from __future__ import annotations

from comm_core.block import Block
```

```python
#| export
class NoCoding(Block):
    "No channel code: the codeword is the bits."

    outputs = ('codeword',)

    def call(
        self,
        bits, # The bits, (B, ..., k)
    ):
        return bits


class HardDecision(Block):
    "No channel code to decode: each bit is the sign of its LLR (1 where positive)."

    outputs = ('bits_hat',)

    def call(
        self,
        llr, # LLRs, log p(1) / p(0), (B, ..., n)
    ):
        return llr > 0
```

```python
import numpy as np
```

```python
bits = np.array([[0, 1, 1, 0]])
assert NoCoding()(bits) is bits
assert HardDecision()(np.array([[-2.0, 0.5, 3.0, -0.1]])).tolist() == [[False, True, True, False]]
```
